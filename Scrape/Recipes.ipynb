```python
def clean(txt):

    txt = txt.lower()

    e_accent = ['é', 'è', 'ê', 'ë']

    plural = plus_s(unwanted_words)
    clean_words = []

    for char in e_accent:
       txt = txt.replace(char, 'e') 

    if '/' in txt:
        splt = txt.split('/')
        txt = ' '.join(splt)

    if '(' in txt and ')' in txt:
        txt = re.sub(r'\([a-z ]*\)', '', txt)

    if re.search(r'\bin\b', txt):
        txt = re.split(r'\bin\b', txt)[0]

    if re.search(r'\blike\b', txt):
            txt = re.split(r'\blike\b', txt)[0]

    if re.search(r'\bfor\b', txt):
        txt = re.split(r'\bfor\b', txt)[0]

    if re.search(r'\bor\b', txt):
        txxt = re.sub(r'[^a-z\s,]', '', txt)
        txxt = txxt.replace('  ', ' ')
        txxt = re.sub(r'\bor\b', ',', txxt)
        splitted = re.split(',', txxt)

        
        for i,j in enumerate(splitted):
            splitted[i] = j.strip()
            word_count = 0

            for word in j.split():
                if word not in unwanted_words and word not in plural:
                    word_count += 1

            if word_count > 0:
                txt = splitted[i]
                
                break
       
           
    if 'substitute' in txt:
        txt = txt.split('substitute')[0]

    txt = re.sub(r'[^a-z\s]', '', txt)
    txt = txt.strip()

    if 'choice' in txt:
        splitt = re.split(r'\bchoice\b', txt)
        splitt = [word.strip() + ' ' for word in splitt]
        txt = splitt[0] + splitt[1].split(' ')[0]

    if 'and or' in txt:
        randn = np.random.randint(2)
        txt = re.split(r'\bandor\b', txt)[randn]

    for word in txt.split():

        if word not in unwanted_words and word not in plural:
            clean_words.append(word)

    for word in clean_words:

        if word + 's' in clean_words:

            clean_words.remove(word + 's')

        while clean_words.count(word) > 1:
            idx = clean_words.index(word, clean_words.index(word) + 1)
            clean_words.pop(idx)
     
    clean_words = ' '.join(clean_words).strip()


    clean_words = clean_words.replace('optional', 'and')

    clean_words = re.split(r'\band\b', clean_words)
    
    if '' in clean_words:
        clean_words.remove('')

    clean_words = [word.strip() for word in clean_words]

    return clean_words
```
